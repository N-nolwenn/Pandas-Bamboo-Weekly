## Read the "defense expenditure" data from the "table 2" into a Pandas df. We want the first table on that sheet, with current prices and exchange rates. 

```python
filename = r"C:\Users\npigeon1\Pandas-Bamboo-Weekly\BW #124 NATO Spending\240617-def-exp-2024-TABLES-en.xlsx"
df = pd.read_excel(filename, 
                   sheet_name="TABLE2",
                   skiprows=9,
                   nrows=33,
                   index_col=0,
                   usecols="C:N",
                   names=range(2013,2025)
                   )
df
```

## Which 5 countries have increased their NATO Spending most as a percentage, in the latest budget cycle (from 2023-2024)? 

```python
df.pct_change(axis='columns')[2024].nlargest()
```

In other words Romania spent 50 percent more on defense in 2024 than in 2023

## Which increased most since 2014, when the data was first tracked?

 "Let's compare each column with the one n columns before it." We can do that with the periods keyword argument, which has a default value of 1, but can be set to anything.

```python
df.pct_change(axis="columns", periods=len(df.columns)-1)[2024].nlargest()
```

## The US spends more on defense than any other NATO country, with Germany and UK in the 2nd and 3rd position. Find how much the US spent (estimated) in 2024.

```python
us_spending = df.loc["United States", 2024]
us_spending
```

```python
df[2024].sort_values(ascending=False).iloc[3:]
```

## Then see how many other NATO countries defense spending, starting with Germany and the UK and continuing from largest to smallest, woul be needed to match US defense spending. That is, if we'd like to say that "the US spends as much as the nnext-largest contributions combined" what is n?

```python
df[2024].sort_values(ascending=False).iloc[3:].cumsum()
```

La somme cumulée est de 506692 < 967707. Which mean that the US defense spending is greater than all of the other NATO members put together. 

```python
df[2024].sort_values(ascending=False).iloc[3:].expanding().sum().loc[lambda s_: s_ >= us_spending]
```
