## Read from "Sheet 1" in the work file into a pandas df. Examine the file to see what values should be considered `NaN. Make the country-name column the index.

```python
import pandas as pd

filename = r"C:\Users\npigeon1\Pandas-Bamboo-Weekly\BW #127 European comparisons\lfsa_ewhun2$defaultview_spreadsheet.xlsx"
df = pd.read_excel(filename, 
                   sheet_name="Sheet 1", # read sheet 1
                   skiprows=14, # skip the first 14 rows
                   nrows=38, # read only the first 39 rows
                   index_col=0, # use the first column as index
                   na_values=[':', 'not available'], # treat NaN values
                   header=None, # no header row
                   names = range(2015, 2025) # assign names to the remaining non index columns
                   )
df
```

let's examine the first two rows to compare EU and Euro. We observe that people in the Euro area globally work less hours than people in Europe as a whole. 

```python
df.iloc[0:2].T
```

## In 2024 (the most recent year of which we have data), which 5 countries worked the greatest and least number of hours ? Which five countries have most increased the number of hours worked (as a percentage) since 2015, and which have decreased ? 

```python
df[2024].sort_values(ascending=False).head(5)  # Top 5 countries with most hours worked in 2024
```

```python
df[2024].sort_values(ascending=True).head(5)  # Top 5 countries with least hours worked in 2024
```

Pour aller plus vite on peut utiliser les fonctions d'aggregations nlargest et nsmallest qui retournent respectivement les 5 plus grandes/ petites valeurs. On utilise agg pour les combiner. 

```python
df[2024].agg(['nsmallest', 'nlargest'])
```

To know which countries have increased and decreased the most (as a percentage) from 2015 to 2024. We can apply similar logic there, but first we need to grab two columns (2015 and 2024), and apply the Pandas window function pct_change across the columns

```python
df[[2015, 2024]].pct_change(axis='columns', fill_method=None)[2024].agg(['nsmallest', 'nlargest'])
```
