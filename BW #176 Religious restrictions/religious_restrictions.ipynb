### Which 10 countries had the greatest improvement (i.e., lower score) in GRI from 2007 to 2022 ? Which 10 countries showed such improvements in the area of SHI ?
On cherche les 10 pays dont le score a le plus baissé entre les deux dates (2007 → 2022).

```python
df.head()
```

```python
gri_afghanistan_2022 = df[(df['Ctry_EditorialName'] == 'Afghanistan') & (df['Question_Year'].dt.year == 2022)]['GRI'].mean()
gri_afghanistan_2022
```

```python
gri_afghanistan_2007 = df[(df['Ctry_EditorialName'] == 'Afghanistan') & (df['Question_Year'].dt.year == 2007)]['GRI'].mean()
gri_afghanistan_2007
```

```python
gri_afghanistan_2022 - gri_afghanistan_2007
```

```python
df_filtered = df[df['Question_Year'].dt.year.isin([2007, 2022])].groupby(['Ctry_EditorialName', 'Question_Year'])[['GRI']].mean().unstack()
df_filtered.columns = df_filtered.columns.droplevel(0)
df_filtered
```

```python
print(type(df_filtered.columns[0]))
```

```python
df_filtered['improvement'] = df_filtered[pd.Timestamp('2007-01-01')] - df_filtered[pd.Timestamp('2022-01-01')]
df_filtered.sort_values(by='improvement', ascending=False).head(10)
```

On commence par filtrer le df avec `loc` sur la colonne de Date puis on utilise la méthode `isin` pour garder uniquement les données entre 2007 et 2022

```python
(
    df
    .assign(Year=df['Question_Year'].dt.year)
    .loc[lambda d: d['Year'].isin([2007, 2022]),
        ['Ctry_EditorialName', 'Year', 'SHI']]
    .pivot_table(index='Ctry_EditorialName', columns='Year', values='SHI', aggfunc='mean')
    .diff(axis='columns')
    [2022]
    .nsmallest(10)
    .round(2)
)
```
